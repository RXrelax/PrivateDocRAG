### 重置到目标提交

#### Git stash 用法

| 命令                                                      | 作用与说明                                                                                                                                                                                                                                                                                                                                                                                               |
|---------------------------------------------------------|----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| **创建/保存**                                                                                          |                                                                                                                                                                                                                                                                                                                                                                                                          |
| `git stash push -m "<msg>"`                            | 将当前 **暂存区**（已 `git add`）和 **工作区** 的修改一起打包存入栈顶，并加上描述 `<msg>`，之后恢复到上一次提交的干净状态。                                                                                                                                                                                                                                                |
| `git stash push -u`                                    | 同上，但 **一并** 保存 **未跟踪文件**（untracked）。等同于 `--include-untracked`。                                                                                                                                                                                                                                                                                                                            |
| `git stash push -a`                                    | 同上，但 **包括** 忽略文件（ignored files）在内的所有文件修改。等同于 `--all`。**谨慎**使用。                                                                                                                                                                                                                                                                                                                |
| `git stash push -p`                                    | 交互式地选取修改（hunks）保存，可精细控制哪些改动进入 stash。等同于 `--patch`。                                                                                                                                                                                                                                                                                                                          |
| `git stash save "<msg>"`                               | 旧版写法（已废弃），等同于 `git stash push -m "<msg>"`。                                                                                                                                                                                                                                                                                                                                                 |
| **查看**                                                                                               |                                                                                                                                                                                                                                                                                                                                                                                                          |
| `git stash list`                                      | 列出所有 stash 条目（栈），最近的在最前，格式如 `stash@{0}: On <branch>: <msg>`。                                                                                                                                                                                                                                                                                                                             |
| `git stash show [stash@{n}]`                          | 展示指定 stash 的简要差异。如需完整补丁：`git stash show -p stash@{1}`。                                                                                                                                                                                                                                                                                                                                    |
| **恢复**                                                                                               |                                                                                                                                                                                                                                                                                                                                                                                                          |
| `git stash apply [stash@{n}]`                         | 将指定 stash 的改动恢复到当前分支的工作区，但不删除 stash 条目（可多次应用同一条）。默认 `stash@{0}`。                                                                                                                                                                                                                                                                                                        |
| `git stash pop [stash@{n}]`                           | 将指定 stash 的改动恢复到工作区，**并** 删除该 stash 条目（出栈）。默认 `stash@{0}`。                                                                                                                                                                                                                                                                                                                        |
| **删除**                                                                                               |                                                                                                                                                                                                                                                                                                                                                                                                          |
| `git stash drop [stash@{n}]`                          | 删除指定的一条 stash 记录，不恢复改动。默认 `stash@{0}`。                                                                                                                                                                                                                                                                                                                                                    |
| `git stash clear`                                     | **清空** 整个 stash 栈，删除所有记录。**谨慎**使用！                                                                                                                                                                                                                                                                                                                                                         |
| **分支相关**                                                                                           |                                                                                                                                                                                                                                                                                                                                                                                                          |
| `git stash branch <new-branch> [stash@{n}]`           | 从指定的 `stash@{n}` 创建一个新分支 `<new-branch>`，切换到此分支并自动恢复该 stash 的改动，同时删除该 stash。等同于：<br>```bash<br>git branch <new-branch>               # 新建分支但不切换<br>git stash apply [stash@{n}]         # 恢复改动<br>git checkout <new-branch>              # 切分支<br>git stash drop [stash@{n}]           # 删除 stash<br>``` |
| **低级/内部命令**                                                                                     |                                                                                                                                                                                                                                                                                                                                                                                                          |
| `git stash show -p [stash@{n}]`                      | 显示指定 stash 的完整差异补丁（patch）。                                                                                                                                                                                                                                                                                                                                                                     |
| `git stash save --keep-index`                        | （旧写法）保存工作区改动，但**保留**已暂存（index）部分不被清理。等同于 `git stash push --keep-index`。                                                                                                                                                                                                                                                                                                      |
| `git stash push --keep-index`                        | 将工作区中除了已暂存内容外的改动 stash 掉，保留暂存区内容，以便进行临时部分提交或检查。                                                                                                                                                                                                                                                                                                                     |
| `git stash save --patch`                             |（旧写法）交互式选择改动 hunk 来 stash，等同于 `git stash push -p`。                                                                                                                                                                                                                                                                                                                                         |
| `git stash create [-m "<msg>"]`                      | 不往栈中“真正”存储，只在本地生成一个临时的 stash 对象，并返回其对象 ID，可配合 `git stash store` 使用。                                                                                                                                                                                                                                                                                                      |
| `git stash store -m "<msg>" <stash@{n}>`             | 将一个由 `git stash create` 生成的 stash 对象，以新消息入栈。                                                                                                                                                                                                                                                                                                                                                |


#### 重置会丢失之后所有未提交的改动，必要时先把改动 stash 起来：

git stash push -m "临时备份"

git checkout master

# 方法一：恢复并保留 stash
git stash apply stash@{0}

# 方法二：恢复并删除 stash
git stash pop stash@{0}

#### 硬重置到目标提交

git reset --hard 68238ce16f4bfff77fb217ad91a14a99e83a46cb

#### 如果这个分支已经推送到远程，还要同步远程

⚠️ 这步会重写远程历史，谨慎使用，确保团队知晓。

git push Github main --force
